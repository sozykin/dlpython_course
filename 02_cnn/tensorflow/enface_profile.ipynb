```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Rescaling
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.utils import image_dataset_from_directory
from tensorflow.keras.preprocessing import image
from tensorflow import data
import matplotlib.pyplot as plt
from pathlib import Path
import zipfile
%matplotlib inline
```

## Подготовка данных

Загружаем данные для обучения

```python
!curl -s -L -o train.zip "https://www.dropbox.com/scl/fi/j1abedjqgbis6k0qb9g6s/train.zip?rlkey=9b8dk24ssrs966tc0d4np9ca9&dl=1"
```

Распаковываем архив с обучающим набором данных

```python
data_path = 'data'
```

```python
path = Path(data_path)

if not path.exists():
    path.mkdir(parents=True, exist_ok=True)
```

```python
with zipfile.ZipFile('train.zip', 'r') as zip_ref:
    zip_ref.extractall(data_path)
```

**Создаем DataSet'ы**

```python
batch_size = 64
image_size = (100, 100)
```

Набор данных для обучения

```python
train_dataset = image_dataset_from_directory(
    "data/train", 
    image_size=image_size, 
    batch_size=batch_size,
    validation_split=0.2,
    subset="training",
    seed=42
)
```

Проверочный набор данных

```python
val_dataset = image_dataset_from_directory(
    "data/train", 
    image_size=image_size, 
    batch_size=batch_size,
    validation_split=0.2,
    subset="validation",
    seed=42
)
```

**Просмотр примеров данных**

```python
class_names = train_dataset.class_names
```

```python
class_names
```

```python
for images, labels in train_dataset.take(1):
  for i in range(9):
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title(class_names[labels[i]])
    plt.axis("off")
```

**Настраиваем производительность работы DataSet'ов**

```python
AUTOTUNE = data.AUTOTUNE
train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.prefetch(buffer_size=AUTOTUNE)
```

## Создаем нейронную сеть

```python
# Создаем последовательную модель
model = Sequential(
    [
        # Предварительная обработка: нормализация
        Rescaling(1./255),
        # Первый сверточный слой
        Conv2D(32, (3, 3), activation='relu'),
        # Первый слой подвыборки
        MaxPooling2D(pool_size=(2, 2)),
        # Слой регуляризации Dropout
        Dropout(0.25),

        # Второй сверточный слой
        Conv2D(64, (3, 3), activation='relu'),
        # Второй слой подвыборки
        MaxPooling2D(pool_size=(2, 2)),
        # Слой регуляризации Dropout
        Dropout(0.25),

        # Слой преобразования данных из 2D представления в плоское
        Flatten(),
        # Полносвязный слой для классификации
        Dense(512, activation='relu'),
        # Слой регуляризации Dropout
        Dropout(0.5),
        # Выходной полносвязный слой
        Dense(1, activation='sigmoid')
    ]
)
```

**Компилируем модель**

```python
model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

## Обучаем нейронную сеть

```python
history = model.fit(train_dataset,
                    epochs=5,
                    validation_data=val_dataset,
                    verbose=1)
```

```python
plt.plot(history.history['accuracy'],
         label='Доля верных ответов на обучающем наборе')
plt.plot(history.history['val_accuracy'],
         label='Доля верных ответов на проверочном наборе')
plt.xlabel('Эпоха обучения')
plt.ylabel('Доля верных ответов')
plt.legend()
plt.show()
```

```python
plt.plot(history.history['loss'],
         label='Ошибка на обучающем наборе')
plt.plot(history.history['val_loss'],
         label='Ошибка на проверочном наборе')
plt.xlabel('Эпоха обучения')
plt.ylabel('Ошибка')
plt.legend()
plt.show()
```

## Сохраняем обученную нейронную сеть

```python
model.save("enface_profile.keras")
```
